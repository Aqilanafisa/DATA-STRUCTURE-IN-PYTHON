```python
# Mengindikasikan pengelompokan data yang disimpan
class Produk:
  # Mendeklarasikan tempat produk yang dibuat
    def __init__(self, id, nama, harga, rating):
        self.id = id # id produk
        self.nama = nama # nama produk
        self.harga = harga # harga produk
        self.rating = rating # rating produk
  # Mendeklarasikan tampilan data produk saat dijalankan
    def __str__(self):
      # Mengembalikan tampilan data produk dalam bentuk string
        return f"{self.nama} (Rp {self.harga:,} | Rating: {self.rating}/5)"
# Mengindikasikan pengelompokan pengurutan data produk
class SorterProduk:
  # Mendefinisikan produk list yang diurutkan
    def __init__(self):
        self.produk_list = []

  # Menambahkan produk dalam list
    def tambah_produk(self, produk):
        self.produk_list.append(produk)
  # Fungsi menampilkan produk dalam list yang dibuat
    def tampilkan_produk(self, judul):
        print(f"\n{judul}")
        print("-" * 50)
        for produk in self.produk_list:
            print(produk)

  # Mendeklarasikan fungsi quick sort untuk mengurutkan harga produk
    def quick_sort_harga(self, low, high):
      # Fungsi kondisi memeriksa pengurutan data
        if low < high:
          # Mengindikasikan posisi pivot dari data harga
            pi = self.partition_harga(low, high)
          # Mengurutkan bagian kiri atau lebih rendah dari pivot
            self.quick_sort_harga(low, pi-1)
          # Mengurutkan bagian kanan atau lebih tinggi dari pivot
            self.quick_sort_harga(pi+1, high)
  # Mendeklarasian pembagian data harga berdasarkan pivot
    def partition_harga(self, low, high):
      # Mengindikasikan produk teratas sebagai pivot
        pivot = self.produk_list[high].harga
      # Mengindikasikan data yang lebih kecil dari pivot
        i = low - 1
      # Fungsi perulangan untuk menelusuri data dari awal sampai pivot
        for j in range(low, high):
          # Kondisi harga produk lebih kecil atau sama dengan pivot
            if self.produk_list[j].harga <= pivot:
              # Mengindikasikan elemen yang lebih kecil
                i += 1
              # Fungsi menentukan posisi elemen data
                self.produk_list[i], self.produk_list[j] = self.produk_list[j], self.produk_list[i]
      # Menempatkan posisi pivot yang sesuai
        self.produk_list[i+1], self.produk_list[high] = self.produk_list[high], self.produk_list[i+1]
      # Mengembalikan indeks pivot
        return i + 1

  # Mendeklarasikan fungsi quick sort untuk mengurutkan rating produk
    def quick_sort_rating(self, low, high):
        # Fungsi kondisi memeriksa pengurutan data
        if low < high:
          # Mengindikasikan posisi pivot dari data rating
            pi = self.partition_rating(low, high)
          # Mengurutkan bagian kiri atau lebih rendah dari pivot
            self.quick_sort_rating(low, pi-1)
          # Mengurutkan bagian kanan atau lebih tinggi dari pivot
            self.quick_sort_rating(pi+1, high)
  # Mendeklarasian pembagian data rating berdasarkan pivot
    def partition_rating(self, low, high):
      # Mengindikasikan produk teratas sebagai pivot
        pivot = self.produk_list[high].rating
      # Mengindikasikan data yang lebih kecil dari pivot
        i = low - 1
      # Fungsi perulangan untuk menelusuri data dari awal sampai pivot
        for j in range(low, high):
          # Kondisi rating produk lebih besar atau sama dengan pivot
            if self.produk_list[j].rating >= pivot:
              # Mengindikasikan elemen yang lebih kecil
                i += 1
              # Fungsi menentukan posisi elemen data
                self.produk_list[i], self.produk_list[j] = self.produk_list[j], self.produk_list[i]
      # Menempatkan posisi pivot yang sesuai
        self.produk_list[i+1], self.produk_list[high] = self.produk_list[high], self.produk_list[i+1]
      # Mengembalikan indeks pivot
        return i + 1


# Eksekusi implementasi kode pengurutan
# Inisialisasi marketplace sorter
sort = SorterProduk()

# Data produk tidak terurut
produk_data = [
    Produk(1, "Lenovo LOQ", 27000000, 4.8),
    Produk(2, "Macbook PRO M5 ", 30000000, 4.9),
    Produk(3, "Axioo Mybook Hype 1", 4500000, 4.7),
    Produk(4, "Advan Soulmate X2", 4000000, 4.5),
    Produk(5, "ASUS Vivobook Go 14", 11000000, 4.6),
]

# Tambahkan semua produk
for produk in produk_data:
    sort.tambah_produk(produk)

# Tampilkan seluruh produk sebelum diurutkan
print("DAFTAR PRODUK AWAL (Belum diurutkan)")
sort.tampilkan_produk("Daftar Produk Awal")


sorter_quick = SorterProduk()
for p in produk_data: sorter_quick.tambah_produk(p)
sorter_quick.quick_sort_rating(0, len(sorter_quick.produk_list)-1)
sorter_quick.tampilkan_produk("Quick Sort - Rating Tertinggi")
```
