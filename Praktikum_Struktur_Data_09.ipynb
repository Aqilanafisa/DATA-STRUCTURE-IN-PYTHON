```python
# Mengindikasikan fungsi Hash Table sederhana
class SimpleHashTable:
  # Mendeklarasikan data yang dibuat dalam ukuran tertentu
  def __init__(self, size):
    self.size = size # Menyimpan data ukuran tabel dalam size
  # Mengindikasikan penyimpanan data hash table sejumlah ukuran yang ditentukan
    self.table = [None] * size
  # Mendeklarasikan fungsi Hash untuk menentukan letak data key
  def hash_function(self, key):
    # Mengembalikan pada nilai hash dari key-nya dengan
    # indeks tetap dalam batas ukuran tabel
    return hash(key) % self.size
  # Mendeklarasikan fungsi untuk memasukkan data dalam Hash Table
  def insert(self, key, value):
    # Mengindikasikan indeks penyimpanan data dengan fungsi Hash
    index = self.hash_function(key)
    # Menyimpan nilai dalam indeks yang sudah dihitung
    self.table[index] = value
  # Mendeklarasikan fungsi pencarian data key
  def search(self, key):
    # Mengindikasikan perhitungan kembali indeks dari key
    index = self.hash_function(key)
    # Mengembalikan nilai yang terdapat dalam indeks
    return self.table[index]


# Implementasi eksekusi code
ht = SimpleHashTable(10)
ht.insert("sandal", 100)
ht.insert("sepatu", 200)
# Menampilkan ukuran dari data yang diinginkan
print(ht.search("sandal"))

print("Nilai untuk 'apple':", ht.search("sandal"))
print("Nilai untuk 'banana':", ht.search("sepatu"))
```

```python
# Mengindikasikan fungsi Hash Table dengan Linear Probing
class LinearProbingHashTable:
  # Mendeklarasikan data yang dibuat dalam ukuran tertentu
  def __init__(self, size):
    self.size = size # Menyimpan data ukuran tabel dalam size
  # Mengindikasikan penyimpanan data hash table sejumlah ukuran yang ditentukan
    self.table = [None] * size
  # Mendeklarasikan fungsi Hash untuk menentukan letak data key
  def hash_function(self, key):
    # Mengembalikan pada nilai hash dari key-nya dengan
    # indeks tetap dalam batas ukuran tabel
    return hash(key) % self.size
  # Mendeklarasikan fungsi untuk memasukkan data dalam Hash Table
  def insert(self, key, value):
    # Mengindikasikan indeks penyimpanan data dengan fungsi Hash
    index = self.hash_function(key)
    # Menyimpan indeks awal untuk memastikan isi dalam tabel
    original_index = index
    # Fungsi perulangan yang terus berjalan selama indeks masih terisi data
    while self.table[index] is not None:
    # Fungsi untuk mencocokkan key pada indeks dengan key yang ingin dimasukkan
      if self.table[index][0] == key:
        # Perulangan dihentikan saat key bernilai sama
          break
      # Mengindikasikan linear pobing saat indeks bertabrakan
      # dengan pindah ke indeks berikutnya
      index = (index + 1) % self.size
      # Memeriksa pencarian dalam indeks yang sama dengan indeks awal
      if index == original_index:
        # Menampilkan Error saat tabel telah penuh
        raise Exception("Tabel sudah penuh")
    # Menyimpan data dalam tipe dat Tuple
    self.table[index] = (key, value)
  # Mendeklarasikan fungsi pencarian data key
  def search(self, key):
    # Mengindikasikan perhitungan kembali indeks dari key
    index = self.hash_function(key)
    # Mengindikasikan tempat penyimpanan data indeks awal
    original_index = index
    # Fungsi perulangan yang terus berjalan selama indeks masih terisi data
    while self.table[index] is not None:
      # Memeriksa data dari indeks dicocokkan dengan key
      if self.table[index] [0] == key:
          # Mengembalikan nilai dari tuple
          return self.table[index] [1]
      # Mengindikasikan pencarian ke indeks lain saat key belum ditemukan
      index = (index + 1) % self.size
      # Kondisi saat indeks sudah kembali ke permulaan
      # Mengindikasikan seluruh indeks data tabel telah diperiksa
      if index == original_index:
        # Menghentikan perulangan
          break
    # Mengembalikan None saat key tidak ditemukan
    return None

# Implementasi eksekusi code
ht = LinearProbingHashTable(10)
ht.insert("apple", 100)
ht.insert("banana", 200)
ht.insert("cherry", 300)

print("Nilai untuk 'apple':", ht.search("apple"))
print("Nilai untuk 'banana':", ht.search("banana"))
print("Nilai untuk 'cherry':", ht.search("cherry"))
```

```python
# Mengindikasikan fungsi Hash Table dengan Chaining
class ChainingHashTable:
  # Mendeklarasikan data yang dibuat dalam ukuran tertentu
  def __init__(self, size):
    self.size = size # Menyimpan data ukuran tabel dalam size
    # Membuat sejumlah data list kosong dalam rentang ukuran tertentu
    self.table = [[] for _ in range(size)]
  # Mendeklarasikan fungsi Hash untuk menentukan letak data key
  def hash_function(self, key):
    # Mengembalikan pada nilai hash dari key-nya dengan
    # indeks tetap dalam batas ukuran tabel
    return hash(key) % self.size
  # Mendeklarasikan fungsi untuk memasukkan data dalam Hash Table
  def insert(self, key, value):
      # Mengindikasikan indeks penyimpanan data dengan fungsi Hash
      index = self.hash_function(key)
      # fungsi perulangan pada list dalam indeks yang diinginkan
      for i, (k, v) in enumerate(self.table[index]):
          # Memeriksa kecocokan key dengan data dalam tabel
          if k == key:
              # Memperbarui value saat key ditemukan
              self.table[index][i] = (key, value)
              # Memberhentikan fungsi saat data telah diperbarui
              return
      # Menambahkan data baru pada tabel saat key belum ditemukan
      self.table[index].append((key, value))
  # Mendeklarasikan fungsi pencarian data key
  def search(self, key):
      # Mengindikasikan perhitungan kembali indeks dari key
      index = self.hash_function(key)
      # Fungsi perulangan seluruh data dalam indeks
      for k, v in self.table[index]:
          # Memeriksa kecocokan key dengan data dalam tabel
          if k == key:
            # Mengembalikan value saat key ditemukan
              return v
      # Mengembalikan None saat key tidak ditemukan
      return None

# Implementasi eksekusi code
ht = ChainingHashTable(5)
ht.insert("apple", 100)
ht.insert("banana", 200)
ht.insert("cherry", 300)

print("Nilai untuk 'apple':", ht.search("apple"))
print("Nilai untuk 'banana':", ht.search("banana"))
print("Nilai untuk 'cherry':", ht.search("cherry"))
```
