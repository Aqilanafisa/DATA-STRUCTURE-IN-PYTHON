```python
# Manual Binary Tree

# Membuat sebuah simpul (node) dalam binary tree
class Node:
  # Mendeklarasikan fungsi lokasi data disimpan
  def __init__(self, data):
    self.data = data # Menympan nilai yang dibuat dalam node
    self.left = None # Percabangan data dalam node ke kiri
    self.right = None # Percabangan data dalam node ke kanan
# Membuat tree secara manual
root = Node(1)
root.left = Node(2)
root.right = Node(3)
root.left.left = Node(4)
root.left.right = Node(5)

# Mendeklarasikan fungsi penelusuran binary tree dengan inorder traversal
def inorder(node):
  # Fungsi kondisi untuk memeriksa node
  if node:
    inorder(node.left) # Penelusuran subtree ke kiri
    print(node.data, end=' ') # Menampilkan node dalam satu baris sejajar
    inorder(node.right) # Penelusuran subtree ke kanan
print("Hasil inorder traversal dari binary tree:")
inorder(root) # Mmmanggil fungsi inorder dari root tree
```

```python
# Binary Search Tree

# Membuat sebuah simpul (node) dalam binary search tree
class Node:
  # Mendeklarasikan fungsi lokasi data disimpan
  def __init__(self, data):
    self.data = data # Menympan nilai yang dibuat dalam node
    self.left = None # Percabangan data dalam node ke kiri
    self.right = None # Percabangan data dalam node ke kanan

# Membuat node BinarySearchTree untuk mengelola operasi BST
class BinarySearchTree:
  def __init__(self):
    self.root = None
  # Fungsi untuk menambahkan data dalam BST
  def insert(self, root, data):
    if root is None: # Fungsi untuk memeriksa node
        return Node(data) # Mengembalikan data dengan membuat node baru
    if data < root.data: # Fungsi untuk memeriksa node baru terhadap root
      root.left = self.insert(root.left, data) # Memasukkan data ke cabang kiri
    else: # Fungsi kondisi lainnya
      root.right = self.insert(root.right, data) # Memasukka data ke cabang kanan
    return root # Mengembalikan node root

# Implementasi eksekusi fungsi Binary Search Tree
bst = BinarySearchTree()
root = None
data_list = [50, 30, 70, 20, 40, 60, 80]
# Fungsi perulangan nilai dalam list
for value in data_list:
  # Memasukkan nilai dalam list ke struktur data bst
  root = bst.insert(root, value)
  print(f"Data {value} berhasil ditambahkan ke BST.")
```

```python
# Traversal pada BST

# Mendeklarasikan fungsi traversal inorder pada tree
def inorder(node):
  # Mengindikasikan fungsi untuk memeriksa node
    if node:
        inorder(node.left) # Penulusuran subtree ke kiri
        print(node.data, end=' ') # Menampilkan node dalam satu baris sejajar
        inorder(node.right) # Penelusuran subtree ke kanan

# Mendeklarasikan fungsi traversal preorder
def preorder(node):
  # Mengindikasikan fungsi untuk memeriksa node
    if node:
      print(node.data, end=' ') # Menampilkan node dalam satu baris sejajar
      preorder(node.left) # Penulusuran subtree ke kiri
      preorder(node.right) # Penelusuran subtree ke kanan

# Mendeklarasikan fungsi traversal postorder
def postorder(node):
    if node:
      postorder(node.left) # Penulusuran subtree ke kiri
      postorder(node.right) # Penelusuran subtree ke kanan
      print(node.data, end=' ') # Menampilkan node dalam satu baris sejajar

print("Inorder Traversal:")
inorder(root)
print("\nPreorder Traversal:")
preorder(root)
print("\nPostorder Traversal:")
postorder(root)
```

```python
# Pencarian Nilai dalam BST

# Mendeklarasikan fungsi pencarian nilai dalam BST
def search(node, key):
  # Memeriksa node dalam dua kondisi
  if node is None or node.data == key:
    return node # Mengembalikan node
  # Memriksa node yang dicari terhadap data saat ini
  if key < node.data:
    # Mengembalikan ke pencarian subtree ke kiri
    return search(node.left, key)
  # Mengembalikan ke pencarian subtree ke kanan
  return search(node.right, key)

# Implementasi eksekusi pencarian
key = 60
result = search(root, key)
# Memeriksa hasil pencarian
if result:
  print(f"{key} ditemukan dalam tree.")
else:
  print(f"{key} tidak ditemukan.")
key = 25
result = search(root, key)
# Memeriksa hasil pencarian
if result:
  print(f"{key} ditemukan dalam tree.")
else:
  print(f"{key} tidak ditemukan.")
```
