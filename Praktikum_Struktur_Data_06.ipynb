```python
# Implementasi Bubble Sort

# Mendeklarasikan fungsi sorting dengan BubbleSort pada Array
def bubble_sort(arr):
  # Mengindikasikan data 'n' sebagai jumlah atau panjangnya data dalam array
  n = len(arr)
  # Fungsi perulangan untuk memeriksa elemen dalam array
  for i in range(n - 1):
    # Fungsi perulangan untuk membandingkan nilai antar elemen
    for j in range(n - i - 1):
      # Fungsi kondisi ketika elemen pertama lebih besar dari elemen selanjutnya
      if arr[j] > arr[j + 1]:
          # memberikan hasil elemen data dalam pengurutan
          arr[j], arr[j + 1] = arr[j + 1], arr[j]
  # Mengembalikan output hasil dalam array terurut
  return arr

# Eksekusi sorting code
data = [64, 34, 25, 12, 22, 11, 90] # Data array yang acak
# Perintah pengurutan data
sorted_data = bubble_sort(data.copy())
print("Data asli:", data)
print("Data setelah Bubble Sort:", sorted_data)
```

```python
# Implementasi Selection Sort

# Mendeklarasikan fungsi sorting dengan SelectionSort pada Array
def selection_sort(arr):
  # Mengindikasikan data 'n' sebagai jumlah atau panjangnya data dalam array
  n = len(arr)
  # Fungsi perulangan untuk memeriksa elemen dalam array
  for i in range(n - 1):
    # Mengindikasikan indeks i sebagai nilai terkecil
    min_index = i
    # Fungsi perulangan untuk membandingkan elemen
    # yang bernilai lebih kecil dari nilai arr[i] pada array tak terurut
    for j in range(i + 1, n):
      # Kondisi elemen j lebih kecil dari elemen minimum pertama
      if arr[j] < arr[min_index]:
          # Mengindikasikan pembaharuan nilai terkecil menjadi j
           min_index = j
    # Memberikan hasil elemen data dalam pengurutan
    arr[i], arr[min_index] = arr[min_index], arr[i]
  # Mengembalikan output hasil dalam array terurut
  return arr

# Eksekusi sorting code
data = [64, 25, 12, 22, 11] # Data array yang acak
# Perintah pengurutan data
sorted_data = selection_sort(data.copy())
print("Data asli:", data)
print("Data setelah Selection Sort:", sorted_data)
```

```python
# Implementasi Insertion Sort

# Mendeklarasikan fungsi sorting dengan InsertionSort pada Array
def insertion_sort(arr):
  # Mengindikasikan data 'n' sebagai jumlah atau panjangnya data dalam array
  n = len(arr)
  # Fungsi perulangan untuk memeriksa setiap elemen dalam data dari indeks 1
  # sampai terakhir, elemen pada indeks pertama (0) sudah terurut
  for i in range(1, n):
    # Menyimpan nilai dalam list key sebagai data terurut
      key = arr[i]
    # Mengindaksikan perbandingan indeks saat ini dengan sebelumnya
      j = i - 1
      # Fungsi perulangan terus berjalan selama dalam batas
      # Kondisi nilai lebih kecil dari key, mencari letak key yang tepat
      while j >= 0 and arr[j] > key:
        # Mengindikasikan elemen lebih besar berada di sebelah kanan
          arr[j + 1] = arr[j]
        # Mengindikasikan elemen dalam indeks dibandingkan
        # dengan elemen sebelumnya
          j -= 1
      # Mengindikasikan key berada di posisi yang benar
      arr[j + 1] = key
  # Mengembalikan output hasil dalam array terurut
  return arr


# Eksekusi sorting code
data = [64, 34, 25, 12, 22, 11, 90] # Data array yang acak
# Perintah pengurutan data
sorted_data = insertion_sort(data.copy())
print("Data asli:", data)
print("Data setelah Insertion Sort:", sorted_data)
```

```python
# Implementasi Merge sort

# Mendeklarasikan fungsi sorting dengan MergeSort pada Array
def merge_sort(arr):
   # Kondisi jumlah elemen lebih dari 1 yang perlu dibagi
    if len(arr) > 1:
      # Mengindikasikan data sebagai titik tengah array
        mid = len(arr) // 2
      # Mengindikasikan bagian kiri array dari awal sampai titik tengah
        left_half = arr[:mid]
      # Mengindikasikan bagian kanan array setelah titik tengah sampai akhir
        right_half = arr[mid:]
      # Perintah pengurutan data bagian kiri secara rekursif
        merge_sort(left_half)
       # Perintah pengurutan data bagian kanan secara rekursif
        merge_sort(right_half)
      # i (menelusuri data bagian kiri)
      # j (menelusuri data bagian kanan)
      # k (menempatkan hasil dalam array)
        i = j = k = 0
      # Fungsi perulangan berlanjut untuk membagi elemen
      # pada bagian kiri dan kanan selama masih ada elemen di kedua sisi
        while i < len(left_half) and j < len(right_half):
          # Kondisi untuk membandingkan data bagian kiri dan kanan
            if left_half[i] < right_half[j]:
              # Kondisi lemen kiri lebih kecil, dimasukan pada key
                arr[k] = left_half[i]
              # Berpindah ke elemen berikutnya pada bagian kiri
                i += 1
            # Kondisi lain
            else:
              # Elemen kanan lebih kecil atau sama, dimasukkan pada key
                arr[k] = right_half[j]
              # Berpindah ke elemen berikutnya pada bagian kanan
                j += 1
            # Berpindah ke bagian iterasi selanjutnya di array utama
            k += 1
      # Fungsi perulangan berlanjut saat masih ada elemen di bagian kiri
        while i < len(left_half):
          # Memasukkan sisa elemen ke array key atau utama
            arr[k] = left_half[i]
          # Menggeser indeks kiri dan posisi key array
            i += 1
            k += 1
      # Fungsi perulangan berlanjut saat masih ada elemen di bagian kanan
        while j < len(right_half):
          # Memasukkan sisa elemen ke array key atau utama
            arr[k] = right_half[j]
           # Menggeser indeks kiri dan posisi key array
            j += 1
            k += 1
    # Mengembalikan output hasil dalam array terurut
    return arr



# Eksekusi sorting code
data = [64, 25, 12, 22, 11] # Data array yang acak
# Perintah pengurutan data
sorted_data = merge_sort(data.copy())
print("Data asli:", data)
print("Data setelah Merge Sort:", sorted_data)

```

```python
# Implementasi Quick Sort

# Mendeklarasikan fungsi sorting dengan QuickSort pada Array
def quick_sort(arr):
   # Mengindikasikan ondisi panjang array sama dengan kurang dari 1
    if len(arr) <= 1:
      # Fungsi mengembalikan pada array, karena sudah terurut
        return arr
  # Mengindikasikan nilai yang menjadi perbandingan data pada array
  # Terletak di tengah data dalam iterasi
    pivot = arr[len(arr) // 2]
  # Mengindikasikan list di sisi kiri pivot yang berarti lebih kecil
    left = [x for x in arr if x < pivot]
  # Mengindikasikan nilai tengah dari iterasinya, isinya sama dengan pivot
    middle = [x for x in arr if x == pivot]
  # Mengindikasikan list di sisi kanan pivot yang berarti lebih besar
    right = [x for x in arr if x > pivot]
  # Mengembalikan nilai dalam kondisi terurut
    return quick_sort(left) + middle + quick_sort(right)


# Eksekusi sorting code
data = [64, 25, 12, 22, 11] # Data array yang acak
# Perintah pengurutan data
sorted_data = quick_sort(data.copy())
print("Data asli:", data)
print("Data setelah Quick Sort:", sorted_data)
```
