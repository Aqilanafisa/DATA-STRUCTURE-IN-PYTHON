```python
# Graf dengan Adjacency List

# Graph tak berarah dengan struktur data dictionary
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E']
}

# Menampilkan graf
# Mendeklarasikan fungsi untuk menampilkan graph adjacency list
def print_graph(graf):
    # Fungsi perulangan setiap data dalam graph
    for node in graf:
        # Menampilkan data node yang terhubung dalam adjacency list
        print(f"{node} -> {graf[node]}")
print_graph(graph)

```

```python
# Traversal Graf menggunakan BFS (Breadth First Search)

# Fungsi untuk mengambil daque dari collections
from collections import deque

# Mendeklarasikan fungsi penelusuruan dengan BFS
def bfs(graph, start):
  # Mengindikasikan variabel yang berisi set kosong
    visited = set()
  # Mengindikasikan antrian dengan deque
    queue = deque([start])
  # Fungsi perulangan yang terus berjalan selama antrian berisi data
    while queue:
      # Mengindikasikan variabel penghapusan data
      # paling kiri (depan) dari antrian
        vertex = queue.popleft()
      # Fungsi untuk memriksa variabel vertex
        if vertex not in visited:
            print(vertex, end=" ") # Menampilkan data dalam satu baris sejajar
            visited.add(vertex) # Menambahkan data dalam variabel
            # Menambahkan tetangga node dalam antrian
            queue.extend([neighbor for neighbor in graph[vertex]
                          if neighbor not in visited])

# Implementasi eksekusi fungsi
bfs(graph, 'A')
```

```python
# Traversal Graf menggunakan DFS (Depth First Node)

# Mendeklarasikan fungsi pencarian graph dengan DFS
def dfs(graph, start, visited=None):
  # Memeriksa node
    if visited is None:
      # Membuat variabel visited berupa set kosong
        visited = set()
  # Memeriksa node dalam kondisi lain
    if start not in visited:
        print(start, end=" ") # Menampilkan data dalam satu baris sejajr
        visited.add(start) # Menambahkan node start dalam visited
        # Fungsi perulangan pada semua tetangga yang terhubung dalam graph
        for neighbor in graph[start]:
          # Memanggil fungsi DFS untuk setiap tetangga
            dfs(graph, neighbor, visited)

# Impleemntasi eksekusi fungsi
dfs(graph, 'A')
```

```python
# Melengkapi fungsi untuk menghitung jumlah tetangga dari tiap simpul dalam graf
def count_neighbors(graph):
    result = {}
    for node in graph:
        # Hitung jumlah tetangga dengan mengambil panjang dari list tetangga
        result[node] = len(graph[node])
    # Mengembalikan hasil perhitungan tetangga
    return result

# Implementasi eksekusi perhitungan tetangga
if __name__ == "__main__":
    # Contoh graf representasi adjacency list
    my_graph = {
        'A': ['B', 'C'],
        'B': ['A', 'D', 'E'],
        'C': ['A', 'F'],
        'D': ['B'],
        'E': ['B', 'F'],
        'F': ['C', 'E']
    }

    # Memanggil fungsi penghitungan
    neighbor_counts = count_neighbors(my_graph)
    print("Jumlah tetangga untuk setiap simpul:")
    # Fungsi perulangan
    for node, count in neighbor_counts.items():
        print(f"Simpul {node}: {count} tetangga")
```
