```python
# Case Coin Change in Algoritma Greedy

# Mendeklarasikan fungsi greedy dalam coin change
def coin_change_greedy(amount, coins):
 # Mengurutkan coin dalam kondisi benar (descending)
  coins.sort(reverse=True)
 # Hasil pengurutan nilai coin dalam sebuah list
  result = []
# Fungsi kondisi memeriksa data dalam koin secara berulang
  for coin in coins:
  # Fungsi perulangan yang terus berjalan selama jumlahnya
  # lebih besar dari data koin atau cukup digunakan untuk ditukar
    while amount >= coin:
      # Mengindikasikan pengurangan koin yang tersisa dengan koin yang dipilih
      amount -= coin
      # Menambahkan koin ke dalam list result
      result.append(coin)
  # Mengembalikan hasil akhir koin yang digunakan
  return result


# Implementasi eksekusi kode
amount = 57
coins = [1, 10, 20, 50, 100, 200]
result = coin_change_greedy(amount, coins)
print("Koin yang digunakan:", result)
```

```python
# Case Fractional Knapsack in Greedy

# Mendeklarasikan Fungsi Fractional Knapsack
def fractional_knapsack(items, capacity):
  # items = list of tuples (value, weight)
  # Mengindikasikan pengurutan berdasarkan rasio nilai terbesar ke terkecil
  items = sorted(items, key=lambda x: x[0] / x[1], reverse=True)
  # Mengindikasikan total nilai yang tersimpan
  total_value = 0.0

 # Fungsi perulangan untuk memeriksa nilai dan berat dari items
  for value, weight in items:
    # Fungsi kondisi kapasitas cukup untuk memuat beban item
    if capacity >= weight:
      # Fungsi menambahkan nilai barang ke total nilai
      total_value += value
      # Mengurangi kapaasitas sesuai berat item
      capacity -= weight
    else:
     # Mengurangi nilai sesuai kapasitas yang ada
      total_value += value * capacity / weight
      break # Menghentikan fungsi perulangan saat sudah penuh
  # Mengembalikan nilai total maksimum kapasitas
  return total_value

# Implementasi eksekusi kode
items = [(60, 10), (100, 20), (120, 30)]
capacity = 50
max_value = fractional_knapsack(items, capacity)
print("Nilai maksimum yang dapat diperoleh:", max_value)
```

```python
# Case Minimum Spanning Tree
# Use Prim and Kruskal in Greedy

# Import modul untuk membuat antrian prioritas (priority queue)
import heapq

print("="*20, "Metode Prim", "="*20)
# Mendeklarasikan fungsi prim dalam simpul
def prim(graph, start):
  # Mengindikasikan tempat simpul yang sudah dikunjungi
  visited = set()
  # Mengindikasikan antrian prioritas
  min_heap = [(0, start)]
  # Menyimpan total bobot MST
  total_weight = 0

# Fungsi perulangan yang terus berjalan selama antrian prioritas terisi
  while min_heap:
    # Mengambil bobot terkecil dari antrian
    weight, node = heapq.heappop(min_heap)
    # Fungsi kondisi saat node telag dikunjungi
    if node in visited:
      # Perulangan berlanjut dan node dilewati
      continue
    # Menambahkan node dalam data list yang telah dikunjungi
    visited.add(node)
    # Menambahkan bobot ke MST
    total_weight += weight

    # Fungsi untuk memeriksa seluruh tetangga node
    for neighbor, edge_weight in graph[node]:
      # Fungsi kondisi tetangga belum dikunjungi
      if neighbor not in visited:
        # Menambahkan node ke dalam antrian prioritas
        heapq.heappush(min_heap, (edge_weight, neighbor))
      # Menambahkan node ke dalam list telah dikunjungi
      visited.add(node)
  # Mengembalikan total bobot MST
  return total_weight

# Eksekusi implementasi dalam graf
graph = {
  'A': [('B', 2), ('C', 3)],
  'B': [('A', 2), ('C', 4), ('D', 5)],
  'C': [('A', 3), ('B', 4), ('D', 6)],
  'D': [('B', 5), ('C', 6)]
}

print ("Total bobot MST (Prim):", prim(graph, 'A'))



print("="*20, "Metode Kruskal", "="*20)

# Mendeklarasikan fungsi Kruskal
def kruskal(edges, n_nodes):
  # Mengindikasikan struktur Union-Find
  parent = {i: i for i in range(n_nodes)}

# Mendeklarasikan fungsi pencarian akar dari suatu node
  def find(x):
    #Fungsi terus berjalan selama node bukan akar
    while parent[x] != x:
      # Node diindakasikan sebagai parent node berikutnya
        x = parent[x]
    # Mengembalikan akar node
    return x

# Mendeklarasikan Fungsi Union untuk menggambungkan dua simpul
  def union(x, y):
    # Mencari akar dari node x
    x_root = find(x)
    # Mencari akar dari node y
    y_root = find(y)
    # Fungsi kondisi saat akar y dan x berbeda
    if x_root != y_root:
      # Menggabungkan akar dalam satu simpul
        parent[y_root] = x_root
      # Mengembalikan simpul yang tergabung
        return True
    # Mengembalikan gagal saat simpul sudah terhimpun menjadi satu
    return False

# Mengurutkan edge berdasarkan bobot
  edges.sort(key=lambda x: x[2])
  # Mengindikasikan total bobot MST tersimpan
  total_weight = 0
  # Fungsi memeriksa seluruh edges
  for u, v, weight in edges:
    # Kondisi edges tidak dalam satu simpul
    if union(u, v):
      # Menambahkan bobor MST
      total_weight += weight
  # Mengembalikan total bobot MST
  return total_weight


# Implementasi eksekusi kode dengan data integer
edges = [(0, 1, 2), (0, 2, 3), (1, 2,1), (1, 3, 1), (2, 3, 6)]
n_nodes = 4
print("Total bobot MST (Kruskal):", kruskal(edges, n_nodes))
```
